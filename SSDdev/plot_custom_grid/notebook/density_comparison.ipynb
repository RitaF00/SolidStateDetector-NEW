```julia
using Pkg

using JSON
using SolidStateDetectors
using Unitful
using Plots
using LegendHDF5IO
```

```julia
grid_density = ["58 x 114",  "84 x 132" , "96 x 212", "110 x 222",    "142 x 202",  "168 x 264",    "336 x 420",  "360 x 454" , "414 x 1010", "416 x 1010" , "470 x 1064" ]

wp = [0.9978,    0.9972 ,   0.990,  0.989,   0.9659,  0.9344, 0.159, 0.123, 0.106, 0.107, 0.08];

wp_new = [0.9978, 0.9972 ,  0.997,   0.998, 0.996,  0.994, 0.995, 0.996, 0.997, 0.998, 1.00 ];
```

```julia
x = 1:length(grid_density)

p = plot(
    x, wp,
    xlabel = "Grid density (r x z)",
    ylabel = "Minimum  Φω",   # simbolo più leggibile
    label = "",
    marker = :circle,
    lw = 2,
    ms = 6,

    xticks = (x, string.(grid_density)),
    xrotation = 45,

    legend = :bottomright,
    legendfontsize = 12,

    guidefontsize = 14,
    tickfontsize = 11,

    grid = true,
    minorgrid = false,

    framestyle = :box,
    size = (900, 650),
    margin = 5Plots.mm
)


hline!(
    p,
    [0.99],
    lw = 2,
    label = "Φ = 0.99"
)

vline!(
    p,
    [x[4]],
    lw = 2,
    ls = :dash,
    label = "ρ threshold"
)

display(p)
```

```julia
x = 1:length(grid_density)

p = plot(
    x, wp,
    xlabel = "Grid density (r x z)",
    ylabel = "Minimum  Φω",   # simbolo più leggibile
    label = "Old code",
    marker = :circle,
    lw = 2,
    ms = 6,

    xticks = (x, string.(grid_density)),
    xrotation = 45,

    legend = :bottomright,
    legendfontsize = 12,

    guidefontsize = 14,
    tickfontsize = 11,

    grid = true,
    minorgrid = false,

    framestyle = :box,
    size = (900, 650),
    margin = 5Plots.mm
)

plot!(
    p,
    x, wp_new,
    label = "New code",
    marker = :circle,
    lw = 2,
    ms = 6
)

hline!(
    p,
    [0.99],
    lw = 2,
    ls = :dash,
    label = "Φ = 0.99"
)

display(p)
```
