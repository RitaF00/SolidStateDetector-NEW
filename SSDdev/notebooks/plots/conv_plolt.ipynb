```julia

function human_time(t)
    if isnan(t)
        return ""
    end

    total_seconds = round(Int, t * 60)

    h = total_seconds ÷ 3600
    m = (total_seconds % 3600) ÷ 60
    s = total_seconds % 60

    if h > 0
        return @sprintf("%dh %02dmin %02ds", h, m, s)
    elseif m > 0
        return @sprintf("%dmin %02ds", m, s)
    else
        return @sprintf("%ds", s)
    end
end
```
